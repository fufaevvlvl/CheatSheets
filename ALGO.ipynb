# Бинарный поиск

O(log(n)) по времени, O(1) по памяти

```python
# точное совпадение

def bin_search(l, r, a, target):
    while l <= r:
        m = (l + r)//2
        if a[m] == target:
            return m
        elif a[m] < target:
            l = m + 1
        else:
            r = m - 1
    return -1

# первый/последний подходящий

def first_good(l, r, good): # Для last good перевернем массив 
    while l < r:
        m = (l + r)//2
        if good(m):
            r = m
        else:
            l = m + 1
    return l

# рекурсивный вариант

def first_good_rec(l, r, good):
    if l == r:
        return l
    m = (l + r)//2
    if good(m):
        return first_good_rec(l, m, good)
    else:
        return first_good_rec(m+1, r, good)
```

# Сортировки

|          |Скорость       | Память |Не меняет порядок правильных элементов|
|----------|---------------|--------|---|
| Quick sort|O(n log n)->O(n^2) отсортирован обратно | O(n log n)| - |
|Merge Sort |      O(n log n)   | O(n) |     + |
|Heap sort  |       O(n log n)  |  O(1)| - |


```python
def quick_sort(arr):
    # Базовый случай: массив из 0 или 1 элемента уже отсортирован
    if len(arr) <= 1:
        return arr
    
    # 1. Выбираем опорный элемент (возьмем средний, чтобы избежать худшего случая)
    pivot = arr[len(arr) // 2]
    
    # 2. Делим на три списка
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    # 3. Рекурсивно сортируем левую и правую части
    return quick_sort(left) + middle + quick_sort(right)

# Пример
print(quick_sort([5, 2, 9, 1, 5, 6]))
```

```python
def merge_sort(arr):
    # Базовый случай
    if len(arr) <= 1:
        return arr
    
    # 1. Делим пополам
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    # 2. Сливаем две отсортированные половинки
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    # Сравниваем первые элементы и забираем меньший
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    # Добавляем остатки (одна из половинок могла закончиться)
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Пример
print(merge_sort([5, 2, 9, 1, 5, 6]))
```

Куча — это двоичное дерево, где у каждого родителя сила больше, чем у его детей.
Вершина (корень) — самый сильный (максимальный элемент).

Это дерево мы храним не в виде узлов, а в обычном массиве по правилу: Левый ребёнок элемента с индексом i находится по адресу 2*i + 1, Правый — 2*i + 2

Два шага heap sort

Шаг А. Построить кучу из любого массива

Исходный массив — это просто дерево, но оно может не быть кучей. Мы исправляем это, начиная с последнего родителя (индекс n//2 - 1) и поднимаясь вверх. Для каждого родителя мы «просеиваем» его вниз, меняя с самым большим ребёнком, пока он не станет больше обоих детей.

Шаг Б. Сортировка (извлечение максимумов)

Теперь мы знаем, что корень — самый большой элемент. Мы меняем корень с последним элементом массива (тем самым «вынимаем» максимум и кладём его в конец, где он уже не мешается). Потом уменьшаем размер кучи на 1 (последний элемент мы больше не трогаем). Но теперь корень может быть слишком маленьким — мы просеиваем его вниз, восстанавливая свойство кучи.

Повторяем это, пока куча не опустеет. В итоге массив становится отсортированным по возрастанию.

```python


def heap_sort(arr):
    n = len(arr)
    
    # 1. Строим max-heap (пирамиду)
    # Начинаем с последнего родителя (n//2 - 1) и идем вверх
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    
    # 2. Извлекаем элементы из кучи по одному
    for i in range(n - 1, 0, -1):
        # Меняем корень (максимум) с последним элементом
        arr[i], arr[0] = arr[0], arr[i]
        # Восстанавливаем кучу для уменьшенного размера
        heapify(arr, i, 0)
    
    return arr

def heapify(arr, n, i):
    """
    Превращает поддерево с корнем i в max-heap.
    n - размер кучи, i - индекс корня поддерева.
    """
    largest = i          # считаем корень наибольшим
    left = 2 * i + 1     # левый ребенок
    right = 2 * i + 2    # правый ребенок
    
    # Если левый ребенок существует и больше корня
    if left < n and arr[left] > arr[largest]:
        largest = left
    
    # Если правый ребенок существует и больше текущего наибольшего
    if right < n and arr[right] > arr[largest]:
        largest = right
    
    # Если наибольший не корень — меняем и рекурсивно чиним дальше
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

# Пример
arr = [5, 2, 9, 1, 5, 6]
heap_sort(arr)
print(arr)
```

# Деревья

```python
## **Обход дерева в глубину (DFS)**

#### 1. **Рекурсивный**

# Pre-order: корень → левый → правый
def dfs_preorder(x):
    if x is None: return
    visit(x)          # обрабатываем корень
    dfs_preorder(x.left)
    dfs_preorder(x.right)

#### 2. **Итеративный (через стек)**
# Pre-order
def dfs_preorder_iterative(root):
    if root is None: return
    stack = [root]
    while stack:
        x = stack.pop()
        visit(x)
        if x.right: stack.append(x.right)  # правый кладем первым
        if x.left:  stack.append(x.left)   # левый — вторым (чтобы вытащился первым)

### **Обход дерева в ширину (BFS)**
# Использует **очередь** (FIFO).

from collections import deque

def bfs(root):
    if root is None: return
    queue = deque([root])
    while queue:
        x = queue.popleft()   # берем первый элемент
        visit(x)
        if x.left:  queue.append(x.left)   # добавляем левого ребенка
        if x.right: queue.append(x.right)  # добавляем правого ребенка

```

**Кратко о разнице**
| **Тип обхода** | **Порядок**          | **Структура данных** | **Применение**                     |
|---------------|----------------------|----------------------|------------------------------------|
| **DFS**       | Глубина → ширина     | Стек (рекурсия/итеративно) | Поиск путей, проверка связности     |
| **BFS**       | Уровень за уровнем   | Очередь              | Поиск кратчайшего пути, уровневый анализ |

```python
### Поиcк всех перестановок https://habr.com/ru/companies/piter/articles/1011832/
from typing import List, Set 

a = [1, 2, 3]

def find_all_permutations(nums: List[int]) -> List[List[int]]:
    res = []
    backtrack(nums, [], set(), res)
    return res

def backtrack(nums: List[int], candidate: List[int], used: Set[int],
              res: List[List[int]]) -> None:
    # Если текущий кандидат является полной перестановкой,
    # добавляем ее в результат.
    if len(candidate) == len(nums):
        res.append(candidate[:])
        return
    for num in nums:
        if num not in used:
            # Добавляем num к текущей перестановке и
            # отмечаем как использованное.
            candidate.append(num)
            used.add(num)
            # Рекурсивно исследуем все ветви с обновленным
            # кандидатом.
            backtrack(nums, candidate, used, res)
            # Возврат назад: отменяем изменения.
            candidate.pop()
            used.remove(num)
    
find_all_permutations(a)
```
